Revenue vs Profit Trend

```python
monthly = df.groupby(['year','month']).agg({
    'purchase_amount':'sum',
    'profit':'sum'
}).reset_index()

# create proper time axis
monthly['year_month'] = monthly['year'].astype(str) + '-' + monthly['month'].astype(str)

import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(monthly['year_month'], monthly['purchase_amount'], label='Revenue')
plt.plot(monthly['year_month'], monthly['profit'], label='Profit')

plt.xticks(rotation=45)
plt.legend()
plt.title("Revenue vs Profit Trend")
plt.show()
```

**Insight:**

Revenue remains relatively stable across months with moderate fluctuations, indicating consistent customer purchasing behavior rather than strong seasonal growth trends.

**Insight:**

Profit follows a similar pattern to revenue but shows smaller margins, indicating that costs and discounts are reducing overall profitability.

### Discount Impact

```python
import seaborn as sns

sns.boxplot(x='discount_applied', y='profit', data=df)
plt.title("Discount Impact on Profit")
plt.show()
```

**Insight**:

Orders with discounts show noticeably lower profit compared to non-discounted orders, indicating that discount strategies may be negatively impacting margins.

### Repeat Customer Analysis

```python
df.groupby('is_repeat_customer')['purchase_amount'].mean()
```

**Insight**:

Repeat customers demonstrate higher average spending compared to new customers, highlighting their importance in driving revenue.

### CLV Distribution

```python
sns.histplot(df['clv'], kde=True)
plt.title("Customer Lifetime Value Distribution")
plt.show()
```

**Insight:**

Customer Lifetime Value is right-skewed, meaning a small proportion of customers contributes a large share of total revenue.

### Profit by Category

```python
df.groupby('category')['profit'].sum().sort_values().plot(kind='bar')
plt.title("Profit by Category")
plt.show()
```

**Insight**:

Profit distribution across categories is uneven, suggesting that certain product categories contribute more effectively to profitability and should be prioritized.

### Acquisition Cost vs Profit

```python
sns.scatterplot(x='acquisition_cost', y='profit', data=df)
plt.title("Acquisition Cost vs Profit")
plt.show()
```

**Insight**:

There is no strong positive relationship between acquisition cost and profit, suggesting inefficiencies in marketing spend.

### Retention Analysis

```python
df['retention_flag'].value_counts(normalize=True) * 100
```

**Insight**:

Around 67% of customers are repeat buyers, indicating strong customer retention and a stable returning customer base.

## Final Business Insights


* Revenue is stable but profitability is impacted by discounts and costs.
* Repeat customers contribute significantly to overall revenue.
* Customer retention (~67%) indicates a loyal customer base.
* High-value customers drive a large portion of business performance.
* Marketing spend shows inefficiencies as acquisition cost does not always  translate into higher profit.
